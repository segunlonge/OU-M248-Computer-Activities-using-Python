##  Computer Book A: Is the uniform model reasonable?

A open ended question for statistical analysis:

The observed data bar chart gives the months of death for 82 descendants of Queen Victoria who all died of natural causes (the raw data is in the table below).



There were more deaths in some months than other months but the question is: are the differences due to random variations (due to sample size of the data) in which case it can be fitted to a uniform discrete model or are the differences due to something else which would make it more likely to die in some months than other months?

By inputing a larger sample size n in the code below it can be seen that the bar chart tends towards a normal distribution. However, by inputing sample size of the observed data (82) the barchart resembles less of a discreet uniform model.

## Computer activity: 32 Royal deaths

```python
import matplotlib.pyplot as plt
import numpy as np

# Set the seed so we get the same "random" results every time
np.random.seed(42)

# 1. Prepare Data for Static Charts
categories = np.arange(1, 13)

# Manual estimates for the 'Observed' chart (Red bars)
observed_probs = [0.16, 0.05, 0.085, 0.125, 0.10, 0.05, 0.06, 0.035, 0.05, 0.11, 0.085, 0.10]

# Uniform model (Purple bars) - each is 1/12
uniform_probs = [1/12] * 12

# 2. Generate Dynamic Data (Sample size n=50)
n_sample = 82
samples = np.random.randint(1, 13, size=n_sample)
counts = np.bincount(samples, minlength=13)[1:] # Count occurrences of 1-12
sample_probs = counts / n_sample

# 3. Create the Figure and Subplots
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# Remove the bottom-right empty space
axes[1, 1].axis('off')

# Helper function to style the axes like the original image
def style_ax(ax, title):
    ax.set_title(title, fontsize=12)
    ax.set_xticks(categories)
    ax.set_ylim(0, 0.25)
    ax.set_yticks([0, 0.05, 0.1, 0.15, 0.2, 0.25])
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

# Top Left: Observed Data
axes[0, 0].bar(categories, observed_probs, color='#c0392b', edgecolor='black', linewidth=0.7)
style_ax(axes[0, 0], 'Bar chart of observed data')

# Bottom Left: Uniform Model
axes[1, 0].bar(categories, uniform_probs, color='#7d3c98', edgecolor='black', linewidth=0.7)
style_ax(axes[1, 0], 'Discrete uniform model')

# Top Right: Dynamic n=50 Chart
axes[0, 1].bar(categories, sample_probs, color='#c0392b', edgecolor='black', linewidth=0.7)
style_ax(axes[0, 1], f'n={n_sample}')
# Bold label for n=50 at the very top as seen in image
axes[0, 1].text(0.5, 1.1, f'n={n_sample}', transform=axes[0, 1].transAxes, 
               ha='center', weight='bold', bbox=dict(facecolor='none', edgecolor='black', boxstyle='round'))

plt.tight_layout()
plt.show()
```

**The larger sample size of n = 5000 resembles more of a discrete uniform distribution model than sample size of the observed data set of n = 82**



#### Commentary

The observed data contains a small sample of just 82. It was been shown than with larger sample sizes the observations would resemble a discrete uniform distribution. Therefore, the overall conclusion is that larger dataset is needed for royal deaths before it can be concluded that it fits a uniform discrete model or not or that the month in which death occurs is the reason for the variability e.g. such as winter months.
