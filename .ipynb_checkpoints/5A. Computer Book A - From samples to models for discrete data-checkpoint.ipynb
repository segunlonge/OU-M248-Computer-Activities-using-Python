## Computer activity 24: The score of a die

The rolling of a 6-sided die has a discrete uniform distribution given by the expression below i.e. for any number on the face of the die from 1 to 6 each has an equal chance of $\frac{1}{6}$



The more times the time is rolled the more it will resemble a perfect uniform distribution e.g. using the simulation below a comparison of 300 rolls versus 3000 rolls makes the point obvious. This is due to several reasons:

1. Equal Probability: As mentioned by the formula above for uniform distribution each face of the die as an equal probability of $\frac{1}{6}$ of landing face up.
2. Small sample size: Small sample size introduce a degree of randomness which would be smoothed out in larger samples or rolls of the die.
3. The Law of Large Numbers: As the sample size or number of rolls beging to increase the randomness begin to cancel out resulting in a near perfect uniform distribution.

```python
import numpy as np
import matplotlib.pyplot as plt

def simulate_die(n):
    # Use randint to simulate the rolling of a 6-sided die
    rolls = np.random.randint(1, 7, size=n)
    
    # Count how many times each number appeared in all the total number of rolls
    faces = np.arange(1, 7) # create an array 1,2,3,4,5,6
    counts = np.array([np.sum(rolls == face) for face in faces]) # for each face of the die count the frequency
    
    # Calculate Relative Frequency (Count / Total Rolls)
    relative_frequencies = counts / n
    # print(relative_frequencies)
    
    # 4. Create the graph
    plt.bar(faces, relative_frequencies, color='firebrick', edgecolor='black')
    plt.ylim(0, 1) # Set the Y-axis from 0 to 1
    plt.xlabel('Score on die')
    plt.ylabel('Relative frequency')
    plt.title(f'Simulation of {n} Rolls')
    plt.show()

# Adjust the sample size here to see the effect of large sample sizes against small sample sizes
sample_size = 3000
simulate_die(sample_size)
```
