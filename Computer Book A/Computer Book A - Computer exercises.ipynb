# 3 Passengers on standby

According to the publicity department of an international airline, on average 10% of people making reservations on their scheduled London–Rome service fail to turn up for their flight. Suppose that on a particular day, the flight is fully booked for 140 passengers with 16 more waiting on standby.

* If the airline’s claim is accurate, what is the distribution of $X$, the number of passengers who do not turn up for the flight?
* Find the probability that all the standby passengers get seats on the flight.
* Find the probability that exactly half of the standby passengers get seats on the flight.

### What is the distribution of (random variable) $X$, the number of passengers who do not turn up for the flight?

Since there are only two possible outcomes 'passenger turns up' or 'passenger does not turn up' and the chances of a single passenger turning up or not for the flight is indepedent from the next passenger. A binomial probability distribution model would seem to be the most appropriate way of modelling the probability distribution of any one of the airline passengers missing their flight.

Hence, the binomial distribution is $X ∼ B(140, 0.1)$

### Find the probability that all the standby passengers get seats on the flight.

For the standby passentgers (16 total) to have a chance of getting a seat on the London-Rome flight **at least** 16 of the 140 passengers who already booked must miss their flights.

Hence, the probability that all standby passengers get seats on the flight is: $P (X \ge 16) = 1 - P (T ≤ 15)$

```python
from scipy.stats import binom

n = 140
p = 0.1
k = 15

# Calculate P(X <= 15)
prob_le_15 = round(float(binom.cdf(k, n, p)),4)

# Calculate P(X >= 16) = 1 - P(X <= 15)
result = round(float(1 - prob_le_15), 4)

print(f"{prob_le_15=}")
print(f"{result=}")
```

### Commentary

The probability of all standby passengers getting on the flight is obviously contigent on 16 passengers out of the fully booked 140 passengers missing their flights. So the probability of at least 16 booked passengers missing their flight is 32.52%

### Find the probability that exactly half of the standby passengers get seats on the flight

The probability of **exactly** half of the passengers on standby i.e. 8 getting a seat on the flight is: $P(X = 8)$

```python
from scipy.stats import binom

n = 140
p = 0.1
k = 8

# Calculate P(X = 8)
prob_e_8 = round(float(binom.pmf(k, n, p)),4)

print(f"{prob_e_8=}")
```

### Commentary

Since we are interested in the exact probability instead of the probability 'upto' or 'greater than' we use the PMF (Probability Mass Function) as opposed to the CDF (Cumulative Distribution Function).

Hence the probability of exactly 8 of the standby passengers getting seats on the flight is: 2.72%

The arrival times of patients at an intensive-care unit were recorded in the period 4 February 1963 to 19 March 1964. The Minitab worksheet intensive care.mwx contains data on the 40 waiting times (to the nearest half hour) between the first 41 admissions. The purpose of collecting the data was to identify any systematic variations in arrival rates.

* Investigate whether or not these data are consistent with an exponential distribution being a good model for the waiting times between admissions.
* The data are ordered. Investigate whether the data are consistent with the admission rate remaining constant over time.

```python
#load data from csv
df = pd.read_csv(r'C:\Data Analysis\OU M248 Data Analysis using Python\Data files\intensive-care.csv')
print(df.head(10)) #return the first 5 rows of the dataframe
```

### Investigate whether or not these data are consistent with an exponential distribution being a good model for the waiting times between admissions

```python
import pandas as pd

# Calculate the Mean
mean_val = df['Interval'].mean()

# Calculate the Standard Deviation
std_val = df['Interval'].std()

# Calculate the Coefficient of Variation (CV)
# For an exponential distribution, this should be close to 1
cv = std_val / mean_val

print(f"Mean: {mean_val:,.4f}")
print(f"Standard Deviation: {std_val:,.4f}")
print(f"Coefficient of Variation: {cv:,.4f}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

#load data from csv
df = pd.read_csv(r'C:\Data Analysis\OU M248 Data Analysis using Python\Data files\intensive-care.csv')
print(df.head(10)) #return the first 5 rows of the dataframe

# Plot histogram from Weight change column
plt.figure(figsize=(8, 6))
plt.hist(df['Interval'], color='steelblue', edgecolor='black', density=False) #Density=False means not to convert to unit area histog

# Add labels and a title
plt.title('Cumulitive arrival times of patients at an intensive-care unit')
plt.xlabel('Interval')
plt.ylabel('Frequency')
```

#### Commentary

An exponential model can be considered for analysing a dataset if the mean and standard deviation are roughly equal and if the frequency plot of the data follows that of an exponential PDF.

In the analysis above the coefficient of variation is close to 1 meaning that the mean and standard deviation can be considered reasonably close. Also, the frequency plot of the dataset has a very close same to an exponential distribution. So an exponential model can be considered appropriate for anaylising the dataset.



### The data are ordered. Investigate whether the data are consistent with the admission rate remaining constant over time.
