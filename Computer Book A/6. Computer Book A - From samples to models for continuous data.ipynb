## Computer activity 26 & 27: The weights of American adults and Modelling weights

The weight of a randomly selected American adult can be modelled by a random variable as demonstrated below. 

As larger and larger samples are taken, the shape of the histogram becomes less jagged, suggesting that a smooth curve might provide an
adequate model for the probability distribution of this random variable (the sample size can be changed by entering a different value of n in the code below).

The model for American adult weights as can be seen is unimodal (similar to a normal distribution) but also right-skewed meaning that more Americans weigh more than the average as compared to those that weight less than the average.

Since weights can take any value inbetween an interval of values (depending on the precision of the weighing machine) a continous model is preferable. Contrast this with finding a model for the random variable (outcome) of a 6 sided die; the values are strictly integers i.e. nothing in between throwing a 5 or a 6. 

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns # Used for the smooth density curve

def plot_unit_area_histogram(n):
    mu, sigma = 4.35, 0.2 
    # mu is the mean of the natural logarithm of the weights and is selected to slightly right-skewed the histogram
    # It is set at 4.35 as that shifts the mean to the left slightly 
    # sigma is 0.2 as it sets the standard deviation to be between 50kg and 120kg where most American adults are likely to lie in terms of weight
    weights = np.random.lognormal(mean=mu, sigma=sigma, size=n) #The lognormal function creates the random weights with a right-side tail
    
    plt.figure(figsize=(8, 6))
    
    # density=True ensures the total area under the bars is 1
    plt.hist(weights, bins=50, range=(0, 200), color='firebrick', 
             edgecolor='black', density=True, alpha=0.7, label=f'Sample (n={n})')
    
    # 3. Add a smooth density curve (KDE)
    # Adding a smooth density curve to the histogram which has a total area of 1
    sns.kdeplot(weights, color='black', linewidth=2, label='Density Curve')
    
    # Formatting the chart
    plt.title(f"Unit-Area Histogram of American Weights (n={n})", fontsize=14)
    plt.xlabel("Weight (kg)")
    plt.ylabel("Probability Density")
    plt.xlim(0, 200)
    plt.ylim(0, 0.04)
    plt.legend()
    plt.grid(axis='y', alpha=0.3)
    
    plt.show()

# choose the sample size by typing it as equal to n
plot_unit_area_histogram(n=5000)
```

### Commentary 

The histogram plotted above is a histogram which shows the relative frequencies of American adult weights taken as a sample. The relative frequency is just a propotion of certain weights compared to the sample total e.g. maybe those weighing 75kg accounting 20 percent of the sample total etc.

For example, to calculate the probability that a randomly selected American adult weighs between 75 and 85 kg then all that is needed is to find the area under the curve between that interval since the total area equals 1. Given the right function for the Density Curve above Calculus (Integration) can be used to get the value of this probability.
